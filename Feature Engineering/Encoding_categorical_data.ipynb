### One hot encoding

```python
cars = pd.read_csv("E:\Data Science\Machine Learning\data\cars.csv")
```

```python
cars
```

```python
for i in cars.columns:
    if (cars[i].dtype) == 'object':
        print(f"unique category {cars[i].unique()}\n")
```

### 1. One Hot encoding using pandas

```python
pd.get_dummies(cars,columns=['fuel','owner'])
```

#### 2. n-1 one hot encoding

```python
pd.get_dummies(cars,columns=['fuel','owner'],drop_first=True)
```

```python
from sklearn.preprocessing import OneHotEncoder

ode = OneHotEncoder(sparse_output=False,dtype=np.int32)

ode.fit_transform(cars[["fuel","owner"]])
```

```python
from sklearn.preprocessing import OneHotEncoder

ode = OneHotEncoder(drop ='first', sparse_output=False,dtype=np.int32)

ode.fit_transform(cars[["fuel","owner"]]).shape
```

```python
from sklearn.preprocessing import OneHotEncoder

ode = OneHotEncoder(sparse_output=False,dtype=np.int32)

ode.fit_transform(cars).shape
```

### One hot encoding with top categories
- when there are huge number of categories we can replace uncommon types with 'other'/'uncommon' etc.

```python
cars.head()
```

```python
counts = cars['brand'].value_counts()
```

```python
threshold = 100

repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(cars['brand'].replace(repl, 'uncommon'))
```
