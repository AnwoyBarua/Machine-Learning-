### Lasso Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import Lasso
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)

plt.figure(figsize= (6,4))
plt.scatter(X,y)

from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

X_draw = np.linspace(-2, 2, 10).reshape(-1,1)
y_draw_pred = lin_reg.predict(X_draw)
plt.plot(X_draw, y_draw_pred, "r-")
plt.show()

### Fitting a linear regression model on the generated data.
```

```python
## Fitting models for different alpha values:

alphas = [0, 0.5, 1, 5, 10, 100]
plt.figure(figsize= (8,6))
plt.scatter(X,y)

for i in alphas:
    lasso_reg = Lasso(alpha= i)
    lasso_reg.fit(X_train, y_train)

    plt.plot(X_test, lasso_reg.predict(X_test), label= f"alpha{i}")

plt.legend()
plt.show()
```

#### 1. Finding how coefficients are effected?

```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
```

```python
data = load_diabetes()

df = pd.DataFrame(data.data,columns=data.feature_names)
df['TARGET'] = data.target

df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(data.data,data.target,test_size=0.2,random_state=42)
```

```python
coefs = []
r2_scores = []

for i in [0,0.1,1,10]:
    reg = Lasso(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())
    y_pred = reg.predict(X_test)
    r2_scores.append(r2_score(y_test,y_pred))
```

```python
plt.figure(figsize=(10, 8))
plt.subplot(221)
plt.bar(data.feature_names,coefs[0])
plt.title('Alpha = 0 ,r2_score = {}'.format(round(r2_scores[0],2)))

plt.subplot(222)
plt.bar(data.feature_names,coefs[1])
plt.title('Alpha = 0.1 ,r2_score = {}'.format(round(r2_scores[1],2)))

plt.subplot(223)
plt.bar(data.feature_names,coefs[2])
plt.title('Alpha = 1 ,r2_score = {}'.format(round(r2_scores[2],2)))

plt.subplot(224)
plt.bar(data.feature_names,coefs[3])
plt.title('Alpha = 10 ,r2_score = {}'.format(round(r2_scores[3],2)))

plt.show()
```

#### Higher the value of alpha more variables are effected.

```python
alphas = [0,0.0001,0.001,0.01,0.1,1,10,100,1000,10000]

coefs = []

for i in alphas:
    reg = Lasso(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())
```

```python
input_array = np.array(coefs)

coefs_df = pd.DataFrame(input_array, columns= data.feature_names)
coefs_df["alpha"] = alphas
coefs_df.set_index("alpha")
# The above output shows coefficients of each variables for different value of alpha.
```

```python
alphas = [0,0.0001,0.0005,0.001,0.005,0.1,0.5,1,5,10]

coefs = []

for i in alphas:
    reg = Lasso(alpha=i)
    reg.fit(X_train,y_train)
    
    coefs.append(reg.coef_.tolist())

input_array = np.array(coefs).T

plt.figure(figsize=(15,8))
plt.plot(alphas,np.zeros(len(alphas)),color='black',linewidth=5)
for i in range(input_array.shape[0]):
    plt.plot(alphas,input_array[i],label=data.feature_names[i])
plt.legend()
```
