```python
import numpy as np

np.random.seed(42)
```

```python
m = 100

X = 2 * np.random.rand(m,1)
y = 4 + 3*X + np.random.randn(m,1)
```

```python
from sklearn.preprocessing import add_dummy_feature

X_b = add_dummy_feature(X)
theta_best = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
```

```python
theta_best
```

##### Original value of intercept and coefficient was 4 and 3, respectively. Using Normal equation formula we have estimated close enough parameters

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(X,y)
lin_reg.intercept_ , lin_reg.coef_
```

##### When XTX is not invertible estimaiting parameters using normal equation is not possible. Then we use another process called pseudoinverse to estimate the parameters

```python
np.linalg.pinv(X_b) @ y
# same answer as above.
```
